# Hungarian ASR error analysis

This notebook analyzes the predictions already committed under `results/`.
It does not run any speech-recognition model, download audio, or change model settings.
Keep validation and held-out test findings separate; the test results are retrospective, not used for model selection.

Run from the repository root or the `notebooks` directory after installing the project.


```python
# Locate the repository from either the root or notebooks directory.
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
repo = next((p for p in (cwd, *cwd.parents) if (p / "pyproject.toml").is_file()
             and (p / "results" / "baselines").is_dir()), None)
if repo is None:
    raise FileNotFoundError("Open this notebook inside the NLP-project checkout")
sys.path.insert(0, str(repo / "scripts"))
sys.path.insert(0, str(repo / "src"))
```

```python
# Rebuild the Markdown report from the saved prediction tables.
from error_analysis import build_report

report_path = repo / "results" / "error_analysis.md"
report = build_report(repo)
report_path.write_text(report, encoding="utf-8")
print(f"Generated {report_path.relative_to(repo)}")
```

```python
# Display the generated tables and reference/hypothesis examples.
try:
    from IPython.display import Markdown, display
except ImportError:
    print(f"Open {report_path} to read the tables.")
else:
    display(Markdown(report))
```
