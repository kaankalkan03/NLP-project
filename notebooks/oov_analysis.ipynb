# Lexical Coverage and OOV Analysis — Hungarian VoxPopuli ASR

This notebook runs the repository's reproducible OOV analysis script and displays its generated report. The training vocabulary is derived from normalized VoxPopuli training references. Test IV/OOV coverage uses exact per-clip token-occurrence overlap; it is descriptive and should not be interpreted as a causal estimate of OOV effects.

**Run from the repository root or from this notebook's directory.** The manifest and test prediction CSVs must be present in the repository.

```python
# Run the analysis script of the repository; it writes results/oov_analysis.md.
from pathlib import Path
import runpy

repo = Path.cwd()
if not (repo / "scripts" / "oov_analysis.py").is_file():
    repo = repo.parent
script = repo / "scripts" / "oov_analysis.py"
if not script.is_file():
    raise FileNotFoundError("Run this notebook from the repository root or notebooks/ directory.")

runpy.run_path(str(script), run_name="__main__")
```

```python
# Show the generated report.
from IPython.display import display, Markdown

report_path = repo / "results" / "oov_analysis.md"
display(Markdown(report_path.read_text(encoding="utf-8")))
```
